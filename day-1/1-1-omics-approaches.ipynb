## Genomic surveillance in vector control

You met this in the [MalariaGEN–PAMCA course](https://anopheles-genomic-surveillance.github.io/home.html),
so here is only a short reminder.

**What.** Genomic surveillance means sampling mosquito populations repeatedly and using DNA or RNA
sequence data to track the genetic changes that matter for control: resistance mutations, species
and population structure, and gene flow between populations.

**Why.** Insecticide-treated nets and indoor residual spraying caused most of the fall in
malaria cases after 2000, and both depend on insecticides still killing mosquitoes. Bioassays
(WHO tubes, CDC bottles) tell you *whether* a population is resistant, but not *why*.
The mechanism matters for choosing a product. For example, resistance driven by cytochrome P450
enzymes can be partly overcome by PBO nets, because PBO inhibits P450s. PBO does nothing for
target-site resistance. Genetic markers can also be screened in dead, stored mosquitoes, in large
numbers and at low cost, so they can act as an early warning before control fails.

**How.** Collect mosquitoes → sequence (whole genomes, RNA, or targeted amplicons) → analyse
→ report to control programmes. The *Anopheles gambiae* 1000 Genomes Project (Ag1000G) [1, 2]
and the MalariaGEN Vector Observatory provide the whole-genome reference data that you used in
the PAMCA course through `malariagen_data`.

**What this course adds.** Whole-genome data tells you *where* variation is, but many resistance
mechanisms act by changing *how much* of a gene is made. We therefore add RNA-seq
(expression), the selection atlas (where selection is acting), assay design (how to test for
a marker cheaply) and amplicon sequencing (how to monitor many markers in many mosquitoes).

## Mechanisms of insecticide resistance

Resistance mechanisms fall into four broad groups (plus behavioural avoidance, which we do not
cover). PAMCA [Workshop 2, Module 2](https://anopheles-genomic-surveillance.github.io/workshop-2/module-2-metabolic-ir.html)
covers metabolic resistance in depth; the table below is a summary.

| Mechanism | What happens | *Anopheles* examples | Main type of genetic change |
|---|---|---|---|
| **Target-site** | The protein the insecticide binds to changes shape, so the insecticide binds less well | *Vgsc* L995F and L995S (*kdr*) for pyrethroids and DDT [3, 4]; *Ace1* G280S for organophosphates and carbamates; *Rdl* A296G for dieldrin | Amino acid substitutions; sometimes gene duplication (*Ace1*) |
| **Metabolic** | Enzymes break down or modify the insecticide before it reaches its target | P450s *CYP6P3*, *CYP6M2*, *CYP6AA1*, *CYP9K1*; glutathione S-transferase *GSTE2*; carboxylesterases *COEAE1F/2F* | More expression (cis-regulatory changes, copy number variants), and sometimes amino acid changes that alter enzyme activity |
| **Cuticular** | A thicker or altered cuticle slows the insecticide getting in | *CYP4G16* and *CYP4G17*, which make cuticular hydrocarbons [5] | Mostly higher expression |
| **Sequestration** | A protein binds the insecticide and stops it acting | Sensory appendage protein **SAP2**, highly expressed in the legs, binds pyrethroids [6] | Higher expression |

Two points to take from this table:

1. **Only target-site resistance is reliably caused by a single amino acid change.** The other
   mechanisms usually work by changing *how much* protein is made. A DNA-only approach finds these
   less easily, which is one reason we need RNA-seq.
2. **The same gene can be affected in several ways at once.** Around Lake Victoria, the
   *Cyp6aa/Cyp6p* cluster on chromosome 2R carries gene duplications, amino acid changes
   (e.g. *Cyp6p4* I236M) and regulatory changes on the same haplotypes. The Ag-vampIR panel you
   analyse on Day 4 tags several of these.

Resistance genes do not act alone. Most resistant populations carry several mechanisms, and
their effects add up (or multiply). That is why surveillance targets panels of markers rather
than one mutation.

## Omic approaches to finding resistance genes

### Candidate gene studies

The oldest approach: pick genes you already suspect (the insecticide's target, detoxification
enzymes) and sequence or measure them. This is how *kdr* was found [3]. It is cheap and the
results are easy to interpret, but **you only find what you already expected**.

### Microarrays, then RNA-seq

Transcriptomics measures the expression of many genes at once in resistant and susceptible
mosquitoes. Early studies used microarrays, first a "detox chip" of about 250 detoxification
genes [7], then whole-genome arrays. These identified *CYP6P3* as overexpressed in
pyrethroid-resistant field mosquitoes, and showed that the enzyme metabolises pyrethroids [8].

RNA-seq replaced microarrays. It measures all genes, including unannotated ones, with a wider
dynamic range. It also produces **sequence**, so the same data can be used to call SNPs, estimate
allele frequencies and compare populations. RNA-Seq-Pop [9] was built to use both parts of the
data, and it is the workflow you run on the Busia data. Because single RNA-seq experiments
often disagree, a **meta-analysis** across many studies (AnoExpress [10]) helps separate
consistent resistance genes from noise.

*Limits:* expression is not causation. A gene can be overexpressed because it sits next to the
causal gene on a swept haplotype, because it is induced by stress, or because the resistant and
susceptible strains differ in many other ways. The susceptible comparator in many studies,
including Busia, is the Kisumu lab strain, which has been in colony for decades.

### Genome-wide association studies (GWAS)

GWAS compares genotypes of mosquitoes that survived or died after insecticide exposure. It links
genotype directly to **phenotype**, which the other approaches do not. A candidate-gene
association study in Ugandan *An. gambiae* found reproducible markers in the *Cyp6aa/Cyp6p*
cluster [11].

*Limits:* bioassay phenotypes are noisy, resistance is polygenic, and a selective sweep makes
long blocks of the genome travel together, so many SNPs are associated and the causal one is
hard to separate. Large sample sizes are needed.

### Selection scans

Strong selection by insecticides leaves footprints in the genome: a stretch of reduced
diversity and long, shared haplotypes around the favoured mutation. Genome-wide selection scans
(H12, G123, iHS; see PAMCA [Workshop 6, Module 3](https://anopheles-genomic-surveillance.github.io/workshop-6/module-3-gwss.html))
find these loci **without any phenotype data**. The malaria vector selection atlas [12] catalogues
signals across thousands of genomes from Ag1000G and the *An. funestus* 1000 Genomes Project.

*Limits:* a signal points to a region, often with several genes in it, and says nothing about
which insecticide (if any) was the selective pressure.

### Copy number variant (CNV) discovery

Gene duplications are a common way to make more enzyme. Whole-genome sequencing reads can be
used to detect CNVs from changes in coverage and from discordant or split reads. Using Ag1000G
data, CNVs were found to cluster at a few resistance loci (*Cyp6aa/Cyp6p*, *Cyp9k1*, the *Gste*
cluster, *Cyp6m/Cyp6z*), often as many independent duplications [13]. A CNV is a strong
explanation for why a gene is overexpressed.

### Amplicon sequencing for monitoring

Once markers are known, you do not need whole genomes to track them. Amplicon sequencing
amplifies tens to hundreds of short targets in one multiplex PCR and sequences them in thousands
of mosquitoes cheaply. **Ag-vampIR** [14] is a panel of this kind, covering resistance markers,
tagging SNPs for the major sweeps, and ancestry-informative markers for species identification.
It can also be used for association studies, as you will do with the Siaya data.

*Limits:* it only sees the markers on the panel, so it monitors rather than discovers.

### Functional validation

None of the approaches above proves that a gene causes resistance. That needs experiments:
expressing the enzyme in cell lines and testing whether it metabolises the insecticide,
knocking the gene down with RNAi, or overexpressing it in transgenic mosquitoes. SAP2 is a
good example where all of these were done [6]. This is outside the scope of the course, but
your dossier verdict should say whether it has been done for your gene.

### Summary of approaches

| Approach | Question it answers | Needs phenotype? | Finds new genes? | Where in this course |
|---|---|---|---|---|
| Candidate gene | Is this suspected gene changed? | Optional | No | Background |
| RNA-seq | Which genes are expressed differently? | Yes (resistant vs susceptible groups) | Yes | Days 1–2 (RNA-Seq-Pop) |
| Meta-analysis | Is the difference consistent across studies? | Yes (in each study) | Yes | Day 2 (AnoExpress) |
| CNV discovery | Is the gene duplicated? | No | Yes | Day 3 (`malariagen_data`) |
| Selection scan | Is the region under recent selection? | No | Yes | Day 3 (selection atlas) |
| GWAS | Is the variant associated with surviving? | Yes (individual dead/alive) | Yes | Day 4 (Siaya, targeted) |
| Amplicon sequencing | How common is the marker, where and when? | Optional | No | Day 4 (AmpSeeker) |
| Functional validation | Does the gene cause resistance? | – | No | Not covered |

## How the approaches fit together: the course study

Each approach has a blind spot, and each one covers a blind spot of another. RNA-seq finds
candidates but cannot separate cause from passenger. Selection scans show that a region matters
in the field but not which gene or why. CNVs and SNPs explain how expression goes up. Amplicon
data links markers to survival in the field. Put together, they make a much stronger case than
any one of them.

This course follows that chain for one public study of pyrethroid resistance in
*An. gambiae* around Lake Victoria:

RNA-seq and selection scans in Busia predicts survival on a deltamethrin net 50 km away in Siaya.

building a gene dossier as you go.

### Exercise 1

A national malaria control programme wants to know, every year, the frequencies of 30 known
resistance markers in about 2,000 mosquitoes from 10 districts. Which approach would you
recommend, and why not whole-genome sequencing?

<details><summary>Show answer</summary>

Amplicon sequencing. The markers are already known, so there is nothing to discover, and a
multiplex panel can genotype thousands of mosquitoes for a small fraction of the cost of whole
genomes. The data are small and quick to analyse, which makes yearly reporting realistic.
Whole-genome sequencing would still be useful for a subset of samples, to look for new signals
the panel cannot see.

### Exercise 2

One of your candidate genes is five times more highly expressed in the deltamethrin-selected
Busia line than in Kisumu. Give three reasons, other than "the gene causes resistance", why this
could happen.

<details><summary>Show answer</summary>

Any three of:

- **Different genetic backgrounds.** Kisumu is an old lab colony from a different place. Many
  genes will differ between the strains for reasons unrelated to insecticides.
- **Hitchhiking.** The gene may sit on the same swept haplotype, or inside the same CNV, as the
  gene that really causes resistance, so it is overexpressed as a side effect.
- **Induction.** If the mosquitoes were exposed to insecticide (or other stress) shortly before
  sampling, detoxification genes can be switched on temporarily.
- **Drift in the colony.** Small lab populations change by chance.
- **Tissue or developmental differences**, such as differences in age or body size between groups.

This is why the course compares selected vs parental Busia lines (same background) as well as
Busia vs Kisumu, and why you check replication in AnoExpress.

</details>

### Exercise 3

The Busia 2016 cohort has a selection signal over the *Cyp6aa/Cyp6p* cluster on 2R, which
contains several P450 genes. What extra evidence would convince you that one particular gene in
the cluster is the target of selection?

<details><summary>Show answer</summary>

Useful lines of evidence include:

- That gene, and not its neighbours, is consistently overexpressed in resistant populations (RNA-seq, AnoExpress).
- A CNV or amino acid change in that gene is at high frequency on the swept haplotype (Ag3 data).
- The signal's focus lies on or next to that gene (selection atlas intervals).
- Markers on the swept haplotype are associated with surviving insecticide exposure (Siaya amplicon data).
- Functional work shows the enzyme metabolises pyrethroids, or knockdown restores susceptibility.

In practice, for gene clusters the answer is often "more than one gene", since duplications can
include several genes.

</details>

## Summary

- Genomic surveillance explains *why* populations are resistant, which guides product choice
  and gives early warning.
- Resistance is target-site, metabolic, cuticular or by sequestration. Apart from target-site
  resistance, most mechanisms work by changing how much protein is made.
- Candidate genes, transcriptomics, GWAS, selection scans, CNV discovery and amplicon sequencing
  each answer a different question and each has blind spots.
- The course follows one chain of evidence: Busia RNA-seq → AnoExpress → Ag3 CNVs and SNPs →
  selection atlas → AnoPrimer → panel design → Siaya amplicons.
